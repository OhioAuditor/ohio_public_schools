## GL_2YR_REV

```python
import pandas as pd


# ============================================================
# MONEY COLUMNS
# ============================================================
money_cols = ["debit", "credit", "balance"]


# ============================================================
# PREPARE REVENUE LOCATION DATA
# ============================================================
def prepare_revenue_location(df):
    gl_2yr_rev_df = df.copy()

    # --------------------------------------------------------
    # Split Fund
    # Example:
    # "Fund: 001 - General"
    # becomes:
    # fund = 001
    # fund_name = General
    # --------------------------------------------------------
    fund_parts = gl_2yr_rev_df["fund"].astype("string").str.extract(
        r"^\s*Fund:\s*(?P<fund>\S+)\s*-\s*(?P<fund_name>.*?)\s*$"
    )

    gl_2yr_rev_df["fund"] = fund_parts["fund"].fillna(
        gl_2yr_rev_df["fund"].astype("string").str.strip()
    )

    gl_2yr_rev_df["fund_name"] = fund_parts["fund_name"]

    # --------------------------------------------------------
    # Split Function
    # Example:
    # "Function: 0000 - Not Applicable"
    # becomes:
    # function = 0000
    # function_name = Not Applicable
    # --------------------------------------------------------
    function_parts = gl_2yr_rev_df["function"].astype("string").str.extract(
        r"^\s*Function:\s*(?P<function>\S+)\s*-\s*(?P<function_name>.*?)\s*$"
    )

    gl_2yr_rev_df["function"] = function_parts["function"].fillna(
        gl_2yr_rev_df["function"].astype("string").str.strip()
    )

    gl_2yr_rev_df["function_name"] = function_parts["function_name"]

    # --------------------------------------------------------
    # Convert money columns to numeric
    # --------------------------------------------------------
    for col in money_cols:
        gl_2yr_rev_df[col] = pd.to_numeric(
            gl_2yr_rev_df[col],
            errors="coerce"
        )

    # --------------------------------------------------------
    # Total each Fund / Function combination
    # --------------------------------------------------------
    gl_2yr_rev_df = (
        gl_2yr_rev_df
        .groupby(
            ["fund", "function"],
            as_index=False,
            dropna=False
        )
        .agg(
            fund_name=("fund_name", "first"),
            function_name=("function_name", "first"),
            debit=(
                "debit",
                lambda values: values.sum(min_count=1)
            ),
            credit=(
                "credit",
                lambda values: values.sum(min_count=1)
            ),
            balance=(
                "balance",
                lambda values: values.sum(min_count=1)
            ),
        )
    )

    return gl_2yr_rev_df


# ============================================================
# PREPARE CURRENT YEAR
# ============================================================
current = prepare_revenue_location(
    rev_loc_df
)


# ============================================================
# PREPARE PREVIOUS YEAR
# ============================================================
previous = prepare_revenue_location(
    rev_loc_prev_df
)


# ============================================================
# MERGE CURRENT AND PREVIOUS YEAR
# ============================================================
rev_2yr_df = current.merge(
    previous,
    on=["fund", "function"],
    how="outer",
    suffixes=("_current", "_prev"),
    validate="one_to_one",
)


# ============================================================
# FUND NAME
# Prefer current-year name.
# If missing, use previous-year name.
# ============================================================
rev_2yr_df["fund_name"] = (
    rev_2yr_df["fund_name_current"]
    .fillna(rev_2yr_df["fund_name_prev"])
)


# ============================================================
# FUNCTION NAME
# Prefer current-year name.
# If missing, use previous-year name.
# ============================================================
rev_2yr_df["function_name"] = (
    rev_2yr_df["function_name_current"]
    .fillna(rev_2yr_df["function_name_prev"])
)


# ============================================================
# VARIANCE
# Current Year Credit - Previous Year Credit
# ============================================================
rev_2yr_df["variance"] = (
    rev_2yr_df["credit_current"]
    - rev_2yr_df["credit_prev"]
)


# ============================================================
# PERCENTAGE CHANGE
#
# (Current - Previous) / Previous * 100
#
# Previous-year amounts of zero are converted to NaN
# to avoid division-by-zero errors.
# ============================================================
rev_2yr_df["percentage"] = (
    rev_2yr_df["variance"]
    .div(
        rev_2yr_df["credit_prev"].replace(
            0,
            float("nan")
        )
    )
    .mul(100)
)


# ============================================================
# FINAL COLUMN ORDER
# ============================================================
rev_2yr_df = rev_2yr_df[
    [
        "fund",
        "fund_name",
        "function",
        "function_name",
        "credit_prev",
        "credit_current",
        "variance",
        "percentage",
    ]
].copy()


# ============================================================
# DISPLAY COPY
#
# Keep rev_2yr_df numeric.
# Only format the copy used for notebook display.
# ============================================================
display_df = rev_2yr_df.head(10).copy()


# Format dollar columns
for col in [
    "credit_prev",
    "credit_current",
    "variance",
]:
    display_df[col] = display_df[col].map(
        lambda value: (
            f"${value:,.2f}"
            if pd.notna(value)
            else ""
        )
    )


# Format percentage column
display_df["percentage"] = display_df["percentage"].map(
    lambda value: (
        f"{value:,.2f}%"
        if pd.notna(value)
        else ""
    )
)


# ============================================================
# DISPLAY RESULTS
# ============================================================
print("Revenue 2-Year Comparison DataFrame:")

display(display_df)
```

## GL_2YR_EXP

```python
money_cols = ["debit", "credit", "balance"]


def prepare_expense_location(df):
    df = df.copy()

    # Split "Fund: 001 - General" into its code and name.
    fund_parts = df["fund"].astype("string").str.extract(
        r"^\s*Fund:\s*(?P<fund>\S+)\s*-\s*(?P<fund_name>.*?)\s*$"
    )
    df["fund"] = fund_parts["fund"].fillna(
        df["fund"].astype("string").str.strip()
    )
    df["fund_name"] = fund_parts["fund_name"]

    # Split "Function: 0000 - Not Applicable" into its code and name.
    function_parts = df["function"].astype("string").str.extract(
        r"^\s*Function:\s*(?P<function>\S+)\s*-\s*(?P<function_name>.*?)\s*$"
    )
    df["function"] = function_parts["function"].fillna(
        df["function"].astype("string").str.strip()
    )
    df["function_name"] = function_parts["function_name"]

    for col in money_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # Total each fund and function combination within the year.
    return (
        df.groupby(["fund", "function"], as_index=False, dropna=False)
        .agg(
            fund_name=("fund_name", "first"),
            function_name=("function_name", "first"),
            debit=("debit", lambda values: values.sum(min_count=1)),
            credit=("credit", lambda values: values.sum(min_count=1)),
            balance=("balance", lambda values: values.sum(min_count=1)),
        )
    )


current = prepare_expense_location(exp_loc_df)
previous = prepare_expense_location(exp_loc_prev_df)

exp_2yr_comparison = current.merge(
    previous,
    on=["fund", "function"],
    how="outer",
    suffixes=("_current", "_prev"),
    validate="one_to_one",
)

# Prefer current-year names; use previous-year names where needed.
exp_2yr_comparison["fund_name"] = (
    exp_2yr_comparison["fund_name_current"]
    .fillna(exp_2yr_comparison["fund_name_prev"])
)
exp_2yr_comparison["function_name"] = (
    exp_2yr_comparison["function_name_current"]
    .fillna(exp_2yr_comparison["function_name_prev"])
)

# Compare current-year debit with previous-year debit.
exp_2yr_comparison["variance"] = (
    exp_2yr_comparison["debit_current"]
    - exp_2yr_comparison["debit_prev"]
)

exp_2yr_comparison["percentage"] = (
    exp_2yr_comparison["variance"]
    .div(exp_2yr_comparison["debit_prev"].replace(0, float("nan")))
    .mul(100)
)

exp_2yr_comparison = exp_2yr_comparison[
    [
        "fund",
        "fund_name",
        "function",
        "function_name",
        "debit_prev",
        "debit_current",
        "variance",
        "percentage",
    ]
]

# Format a copy for display without requiring jinja2.
display_df = exp_2yr_comparison.head(10).copy()

for col in ["debit_prev", "debit_current", "variance"]:
    display_df[col] = display_df[col].map(
        lambda value: f"${value:,.2f}" if pd.notna(value) else ""
    )

display_df["percentage"] = display_df["percentage"].map(
    lambda value: f"{value:,.2f}%" if pd.notna(value) else ""
)

print("Expense Location Comparison DataFrame:")
display(display_df)
```

## GL_Gaps

```python
checks = exp_loc_df[["fund", "function", "ck_no"]].copy()
checks["ck_no"] = pd.to_numeric(
    checks["ck_no"], errors="coerce"
)

checks = (
    checks.dropna(subset=["ck_no"])
    .drop_duplicates()
    .sort_values(["fund", "function", "ck_no"])
)

checks["gap_start"] = checks.groupby(
    ["fund", "function"]
)["ck_no"].shift()

checks["gap_end"] = checks["ck_no"]
checks["missing"] = checks["gap_end"] - checks["gap_start"] - 1

check_number_gaps_df = (
    checks.loc[
        checks["missing"] > 0,
        ["gap_start", "gap_end", "missing"]
    ]
    .astype("Int64")
    .reset_index(drop=True)
)

display(check_number_gaps_df.head(10))
```

```python
def add_filters_and_autofit(tab, verbose):
    """
    Add filters to the tab and autofit the columns.
    
    Parameters:
        tab (xlwings sheet): The tab to add filters and autofit.
        verbose (str): 't' or 'f' to print the number of records in each dataframe.
    Returns:
        None
    """
    # find the last column and row
    last_col = tab.range('A1').expand('table').last_cell.column
    last_row = tab.range('A1').expand('table').last_cell.row
    if verbose == 't':
        print(f"\tLast Column: {last_col}, Last Row: {last_row}")

    # add filters from A1 to last_col and last_row
    tab.range((1, 1), (last_row, last_col)).api.AutoFilter(Field=1)
    # autofit the columns
    tab.autofit()

    # set column width to 50 for any columns that are wider than 50
    for col in range(1, last_col + 1):
        if tab.range(1, col).column_width > 50:
            tab.range(1, col).column_width = 50

    return None
```
